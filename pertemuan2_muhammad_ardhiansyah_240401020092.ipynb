```python
import pandas as pd
import numpy as np

# Muat Dataset Titanic
URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)

# mengecek dataset sudah terimport
print("--- 5 Baris Pertama Dataset Titanic ---")
display(df.head())
```

```python
# --- JAWABAN Q1 ---
print(f"Jumlah Baris : {df.shape[0]}") #untuk menghitung baris
print(f"Jumlah Kolom : {df.shape[1]}") #untuk menghitung kolom
print("\nDaftar Kolom yang Tersedia:")
print(df.columns.tolist()) #untuk melist kolom yang tersedia

#
```

```python
# --- JAWABAN Q2 ---
print("--- Analisis Nilai yang Hilang (Missing Values) ---")
total_missing = df.isnull().sum() #menghitung semua missing values
proporsi_missing = (df.isnull().mean() * 100).round(2) #menghitung persentasenya

# Gabungkan hasil total dan persen ke dalam satu tabel DataFrame agar rapi
missing_table = pd.DataFrame({
    'Total Missing': total_missing,
    'Proporsi (%)': proporsi_missing
})

# Tampilkan kolom yang hanya memiliki nilai missing > 0
print(missing_table[missing_table['Total Missing'] > 0])
```

```python
# --- JAWABAN Q3 ---
# Kolom 'Survived' berisi 1 (Selamat) dan 0 (Meninggal).
# Rata-rata dari kolom ini dikali 100 akan menghasilkan persentase keselamatan.
tingkat_selamat = df['Survived'].mean() * 100

print(f"Persentase Penumpang yang Selamat: {tingkat_selamat:.2f}%")
```

```python
# --- JAWABAN Q4 ---
# Filter menggunakan kondisi ganda (Kondisi 1: Sex == female, Kondisi 2: Pclass == 1)
wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

jumlah_wanita_k1 = len(wanita_kelas1)
rata_usia_wanita_k1 = wanita_kelas1['Age'].mean()

print(f"Jumlah penumpang wanita Kelas 1 : {jumlah_wanita_k1} orang")
print(f"Rata-rata usia mereka           : {rata_usia_wanita_k1:.2f} tahun")
```

```python
# --- JAWABAN Q5 ---
print("--- Analisis Keselamatan Berdasarkan Kelas Penumpang (Pclass) ---")

analisis_kelas = df.groupby('Pclass').agg(
    total_penumpang = ('PassengerId', 'count'),
    jumlah_selamat = ('Survived', 'sum'),
    tingkat_selamat = ('Survived', 'mean')
)

# Konversi tingkat keselamatan ke bentuk persen (%)
analisis_kelas['tingkat_selamat (%)'] = (analisis_kelas['tingkat_selamat'] * 100).round(2)
display(analisis_kelas[['total_penumpang', 'jumlah_selamat', 'tingkat_selamat (%)']])
```

```python
# --- JAWABAN Q6 ---
print("--- Analisis Multivariat: Hubungan Pclass & Sex Terhadap Keselamatan ---")

# Melakukan groupby silang menggunakan dua variabel sekaligus
analisis_multivariat = df.groupby(['Pclass', 'Sex']).agg(
    total_penumpang = ('PassengerId', 'count'),
    jumlah_selamat = ('Survived', 'sum'),
    tingkat_selamat = ('Survived', 'mean')
)

# Konversi ke format persentase
analisis_multivariat['tingkat_selamat (%)'] = (analisis_multivariat['tingkat_selamat'] * 100).round(2)
display(analisis_multivariat[['total_penumpang', 'jumlah_selamat', 'tingkat_selamat (%)']])

# --- Interpretasi Analisis ---
print("\n[Interpretasi Temuan Q6]:")
print("1. Variabel 'Sex' (Jenis Kelamin) merupakan faktor paling dominan yang mempengaruhi keselamatan.")
print("   Meskipun seorang penumpang berada di Kelas 3 (kelas terendah), jika dia seorang wanita,")
print("   tingkat keselamatannya mencapai 50.00%, jauh lebih tinggi dibandingkan pria di Kelas 1 (36.89%).")
print("2. Kombinasi 'kabin mewah' dan 'jenis kelamin wanita' memberikan peluang hidup mutlak.")
print("   Penumpang Wanita di Kelas 1 memiliki tingkat keselamatan fantastis sebesar 96.81% (hanya 3 orang meninggal).")
print("3. Sebaliknya, kelompok dengan tingkat keselamatan paling tragis adalah Pria di Kelas 3,")
print("   di mana hanya 13.54% dari total 347 penumpang pria kelas 3 yang berhasil bertahan hidup.")
```
